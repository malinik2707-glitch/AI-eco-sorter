```python
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image
import numpy as np


class_names = ["bandage", "paper", "metal", "glass", "organic"]  # change if different


img_path = r"medical waste/bandages.jfif"
img = image.load_img(img_path, target_size=(224,224))

plt.imshow(img)
plt.axis('off')
plt.title("Input Image for ECO Sorter")
plt.show()

import tensorflow as tf

model = tf.keras.models.load_model(r"C:\Users\ADMIN\Desktop\eco_sorter\dataset\eco_sorter_model.h5")  

img_array = image.img_to_array(img) / 255.0
img_array = np.expand_dims(img_array, axis=0)


pred = model.predict(img_array)
predicted_class = np.argmax(pred)
confidence = np.max(pred) * 100


print("----- ECO SORTER FINAL OUTPUT -----")
print(f"Input Image: {img_path}")
print(f"Predicted Class: {class_names[predicted_class]}")
print(f"Confidence: {confidence:.2f}%")
print(f"Action: Place the image in the '{class_names[predicted_class]}' bin")
print("-----------------------------------")
```
